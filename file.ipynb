```python

pip install pandas numpy lxml tabulate
import pandas as pd
import numpy as np
from tabulate import tabulate

url = "https://web.archive.org/web/20230902185326/https://en.wikipedia.org/wiki/List_of_countries_by_GDP_%28nominal%29"
CsvFile = pd.read_html(url)
# print(CsvFile[3])
Table = CsvFile[3];
# print(Table.to_string(index=False));
ExtractedTable = Table.iloc[:, 0:3]
# print(ExtractedTable)
Table = ExtractedTable.drop(columns=['UN region'])

Table = Table.drop(index=0)
# print(Table.columns)

Table[('IMF[1][13]', 'Estimate')] = pd.to_numeric(
    Table[('IMF[1][13]', 'Estimate')], errors='coerce'
)


Table = Table.sort_values(
    by=('IMF[1][13]', 'Estimate'),
    ascending=False
)

# print(Table)
Table[('IMF[1][13]', 'Estimate')] = pd.to_numeric(Table[('IMF[1][13]', 'Estimate')], errors='coerce')

Table = Table.dropna(subset=[('IMF[1][13]', 'Estimate')])


Top10 = Table.head(10)



gdp_column = ('IMF[1][13]', 'Estimate')

Top10[gdp_column] = (
    Top10[gdp_column] / 1000
).round(2)

Top10.columns = ['Country', 'GDP (Billion USD)']

print(tabulate(
    Top10,
    headers='keys',
    tablefmt='grid',
    showindex=False
))
```
